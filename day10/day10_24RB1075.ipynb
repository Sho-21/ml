### 24RB1075 小嶋将太郎

```python
import numpy as np
import matplotlib.pyplot as plt
import japanize_matplotlib
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
```

```python
digits = load_digits()
X = digits.data
y = digits.target

X_mean = np.mean(X, axis=0)
X_centered = X - X_mean
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_centered)
```

```python
plt.figure(figsize=(6,5))

for i in range(10):
    plt.scatter(X_pca[y==i, 0],X_pca[y==i, 1], label=str(i), s=10)

plt.title("PCA")
plt.legend()
plt.show()
```

```python
A = np.random.normal(loc=0, scale=np.sqrt(1/2), size=(2, 64))
X_random = X_centered @ A.T

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for i in range(10):
    axes[0].scatter(X_pca[y==i, 0], X_pca[y==i, 1], s=10, label=str(i))

axes[0].set_title("PCA")

for i in range(10):
    axes[1].scatter(X_random[y==i, 0], X_random[y==i, 1], s=10, label=str(i))

axes[1].set_title("ランダム射影")

plt.tight_layout()
plt.show()
```

### 考察
PCAではデータの分散が最大となる方向へ射影するため、同じラベルのデータがまとまって配置され、クラスタ構造が比較的明確に確認できた。一方で、ランダム射影では射影方向をランダムに決めるため、ラベルごとのクラスタが大きく重なってしまい、分離がうまく行えていなかった。この比較から、PCAの方がデータの特徴を保って2次元に圧縮できていると考えられる。
